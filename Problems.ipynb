```python
#how many duplicates in list
lst=list(map(int,input().split()))
count=len(lst)-len(set(lst))
print(count)
```

```python
#how many number are having duplicates
lst=list(map(int,input().split()))
lst2=[]
for i in range(len(lst)):
    if lst.count(lst[i])>1:
        lst2.append(lst[i])
print(set(lst2))

```

```python
# non repeating elements in array
lst=list(map(int,input().split()))
lst2=[]
for i in range(len(lst)):
    if lst.count(lst[i])==1:
        lst2.append(lst[i])
print(lst2)
```

```python
# print missing number between 1 to n
N=5
lst=list(map(int,input().split()))
for i in range(1,N+1):
    if i not in lst:
        print(i)
```
