```python
#Set Operations
s1={1,2,3,4,5,6}
s2={10,22,3,4,6}
print(s1 | s2)
print(s1 & s2)
print(s1 - s2)
print(s1 ^ s2)
```

```python
#Set Methods
s1.add(20)
print(s1)
s1.update((77,88))
print(s1)
s1.discard(2)
s1.pop()
print(s1)
s1.union(s2)
print(s1)
s1.intersection(s2)
print(s1)
s1.difference(s2)
print(s1)
s1.symmetric_difference(s2)
print(s1)
s1.clear()
print(s1)
```
