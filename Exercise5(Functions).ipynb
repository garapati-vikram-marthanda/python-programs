```python
def reverse(num):
  sum=0
  while(num!=0):
    rem=num%10
    sum=sum*10+rem
    num=num//10
  return sum
number = int(input())
print(reverse(number))
```

```python
#function that returns multiple values
def mul_func(*nums):
  s=sum(nums)
  c=len(nums)
  maxi=max(nums)
  mini=min(nums)
  medium=(maxi+mini)/2
  avg=s/c
  return s,c,medium,avg

res_sum,res_count,res_medium,res_avg = mul_func(1,2,33,54,65,7)
print(res_sum)
print(res_count)
print(res_medium)
print(res_avg)
```
