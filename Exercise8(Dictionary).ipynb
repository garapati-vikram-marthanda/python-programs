```python
#Dictionary-Methods
d = {
    "name": "Alice",
    "age": 25,
    "city": "Delhi"
}
print(d.keys())
print(d.values())
print(d.items())
print(d.get("name"))
d.update({"age": 26, "country": "India"})
print(d)
print(len(d))
print(d)
```

```python
#Dictionary-Functions
student = {
    "name": "Rahul",
    "age": 20,
    "marks": 85
}

print(len(student))
print(type(student))


print(max(student))
print(min(student))

print(list(student))
print(tuple(student))
print(set(student))

print("name" in student)
print("city" in student)

print(any(student))
print(all(student))
```

```python
#write a program to check if a given key exist in dictionary or not
dict={
    "name":"Vikram",
    "age":18,
    "city":"Palakol",
    "country":"India"
}
ukey=input()
if(ukey in dict):
    print("Key exist")
else:
    print("Key does not exist")
```

```python
#write a program to add a key,value pair in existing dictionary
dict1={
    "name":"Vikram",
    "age":18,
    "city":"Palakol",
    "country":"India"
}
ukey1=input()
uvalue=input()
dict1[ukey1]=uvalue
print(dict1);
```

```python
#sum all the items in a dictonary
dict2={
    "a":10,
    "b":20,
    "c":30
}
sum=0
for key in dict2:
    sum=sum+dict2[key]
print(sum)
```

```python
#write a program that prints freq or occurence of each element in a list in dictionary format
lst=[1,1,2,2,7,9,5,7,8,9]
freq={}
for i in lst:
  if i in freq:
    freq.update({i:freq[i]+1})
  else:
    freq[i]=1
print(freq)
```

```python
# write a program to print freq of character
str="pythonprogram"
freq={}
for i in str:
  if i in freq:
    freq.update({i:freq[i]+1})
  else:
    freq[i]=1
print(freq)
```
